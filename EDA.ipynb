```python
from torchvision import datasets, transforms
import matplotlib.pyplot as plt
```

```python
train_dataset = datasets.MNIST(
    root="./DATA",
    train=True,
    download=True)
```

```python
train_dataset_digits = [[] for i in range(10)]
for image, label in train_dataset:
    train_dataset_digits[label].append(image)
```

```python
rows = 10
cols = 10
fig, axes = plt.subplots(rows, cols, figsize=(5 * cols, 4 * rows))
for i in range(rows):
    for j in range(cols):
        image = train_dataset_digits[i][j]
        axes[i, j].imshow(image, cmap="gray")
plt.show()        
```
